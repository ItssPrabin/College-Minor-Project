```python
import pandas as pd
#loading final output
df = pd.read_csv("/Users/prabinrimal/Desktop/New>>>>>/text extraction/resumes_extracted.csv")
df
```

```python
# drop duplicate rows
df = df.drop_duplicates()

# drop rows with missing or empty resume_text
df = df.dropna(subset=['resume_text'])
df = df[df['resume_text'].astype(str).str.strip() != '']

# drop rows where resume_text (after stripping) is shorter than 50 characters
df = df[df['resume_text'].astype(str).str.strip().str.len() >= 50]

# reset index and show result
df = df.reset_index(drop=True)
df
```

```python
df = df[["resume_id", "resume_text", "category"]]
df.to_csv("final_dataset.csv", index=False)
df
```
